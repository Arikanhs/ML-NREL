```python
# Data
methods = ['Artificial Neural Network', 'Single-variate LSTM', 'Multi-variate LSTM']
training_r2 = [0.983, 0.976, 0.983]
validation_r2 = [0.972, 0.976, 0.981]
test_r2 = [0.967, 0.962, 0.973]

# Plotting
plt.figure(figsize=(10, 6))

bar_width = 0.3
index = range(len(methods))

plt.bar(index, training_r2, bar_width, label='Training', color='b')
plt.bar([i + bar_width for i in index], validation_r2, bar_width, label='Validation', color='g')
plt.bar([i + 2 * bar_width for i in index], test_r2, bar_width, label='Test', color='r')

# Add labels and title
plt.xlabel('Models')
plt.ylabel('R^2 Values')
plt.title('R^2 Values of Different Models for Training, Validation, and Test Sets')
plt.xticks([i + bar_width for i in index], methods, rotation=0)
plt.legend()

# Add text annotations
for i in index:
    plt.text(i - 0.05, training_r2[i] + 0.01, str(training_r2[i]), fontsize=9)
    plt.text(i + bar_width - 0.05, validation_r2[i] + 0.01, str(validation_r2[i]), fontsize=9)
    plt.text(i + 2 * bar_width - 0.05, test_r2[i] + 0.01, str(test_r2[i]), fontsize=9)

# Show plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# First Dataset "Las Vegas Office"
methods = ['LR', 'ARIMA' , 'SVM', 'KNN', 'RF', 'ANN', 'LSTM-SV', 'LSTM-MV']
mae_values = [78.265, 85.68 , 20.961, 9.63, 10.145, 9.937, 7.48, 6.22]
r_squared_values1 = [0.388, 0.459 ,0.935, 0.950, 0.970, 0.967, 0.962, 0.973]

# Second Dataset "NREL Cafe"
methods2 = ['LR', 'ARIMA' , 'SVM', 'KNN', 'RF', 'ANN', 'LSTM-SV', 'LSTM-MV']
mae_values2 = [5.182, 28.861 , 1.985, 3.866, 2.273, 2.238, 2.347, 2.496]
r_squared_values2 = [0.366, 0.0, 0.828, 0.491, 0.815, 0.81, 0.765, 0.743]

# Third Dataset "Cinemark Simulation"
methods3 = ['LR', 'ARIMA' , 'SVM', 'KNN', 'RF', 'ANN', 'LSTM-SV', 'LSTM-MV']
mae_values3 = [30.822, 161.74 , 9.067, 5.529, 3.929, 5.614, 1.813, 4.209]
r_squared_values3 = [0.882, 0.0, 0.988, 0.980, 0.992, 0.989, 0.989, 0.964]

# Create a figure and axis objects
fig, ax1 = plt.subplots(figsize=(12, 8))

# Plot MAE values for the first dataset
color1 = 'tab:blue'
ax1.set_xlabel('Algorithm')
ax1.set_ylabel('MAE')
ax1.plot(methods, mae_values, marker='o', markersize=8, linestyle='-', color=color1, markerfacecolor='none', markeredgewidth=2, label='Las Vegas Office - MAE')
ax1.tick_params(axis='y', labelcolor=color1)
ax1.set_ylim(0, 90)  # Adjust the range for MAE values

# Plot R-squared values for the first dataset on the secondary y-axis
ax2 = ax1.twinx()
color2 = 'tab:blue'
ax2.set_ylabel('R-squared (R^2)')
ax2.plot(methods, r_squared_values1, marker='o', markersize=8, linestyle='-', color=color2, label='Las Vegas Office - R^2')
ax2.tick_params(axis='y', labelcolor=color2)
ax2.set_ylim(0, 1.05)  # Adjust the range for R-squared values

# Plot MAE values for the second dataset
color3 = 'tab:red'
ax1.plot(methods2, mae_values2, marker='s', markersize=8, linestyle='-', color=color3, markerfacecolor='none', markeredgewidth=2, label='NREL Cafe - MAE')
ax1.set_ylim(0, 90)  # Adjust the range for MAE values

# Plot R-squared values for the second dataset on the secondary y-axis
color4 = 'tab:red'
ax2.plot(methods2, r_squared_values2, marker='s', markersize=8, linestyle='-', color=color4, label='NREL Cafe - R^2')
ax2.set_ylim(0, 1.05)  # Adjust the range for R-squared values

# Plot MAE values for the third dataset
color5 = 'tab:green'
ax1.plot(methods3, mae_values3, marker='v', markersize=8, linestyle='-', color=color5, markerfacecolor='none', markeredgewidth=2, label='Cinemark Simulation - MAE')
ax1.set_ylim(0, 90)  # Adjust the range for MAE values

# Plot R-squared values for the third dataset on the secondary y-axis
color6 = 'tab:green'
ax2.plot(methods3, r_squared_values3, marker='v', markersize=8, linestyle='-', color=color6, label='Cinemark Simulation - R^2')
ax2.set_ylim(0, 1.05)  # Adjust the range for R-squared values

# Set the title
plt.title('Comparison of Algorithms on MAE and R-squared (R^2) for Multiple Datasets')

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, fontsize=12)

# Increase fontsize of y-axis ticks
plt.yticks(fontsize=12)

# Show the legend
# ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

# Place the legend outside of the plot
ax1.legend(loc='center left', bbox_to_anchor=(1.05, 0.90))
ax2.legend(loc='center left', bbox_to_anchor=(1.05, 0.75))

ax1.tick_params(axis='y', colors='black')
ax2.tick_params(axis='y', colors='black')

# Add grid lines with transparency
ax1.grid(alpha=0.25)
# ax2.grid(alpha=0.25)

# Show the plot
# plt.tight_layout()
plt.show()
```
