```python
import pandas as pd 
import numpy as np

from matplotlib import pyplot as plt
from sklearn.preprocessing import StandardScaler

import ipympl 
%matplotlib ipympl


file_path = 'LSTM_2022_predictions.csv'
file_path2 = 'Cafe_2022_for_LSTM.csv'

df = pd.read_csv(file_path)
df_time = pd.read_csv(file_path2)

df['ts'] = df_time['ts']
df.set_index('ts', inplace=True)
df.index = pd.to_datetime(df.index)

plt.figure(figsize=(22,10))

plt.plot(df["Actuals"], label="Real kW", color="blue")
plt.plot(df["Predictions"], label="Predicted kW", color="red")
plt.grid(True)

# Adding labels and title
plt.xlabel("Index")
plt.ylabel("kW")
plt.title("Real kW vs Predicted kW")

# Adding legend
plt.legend()

# Displaying the plot
plt.show()
```
