```python
import pandas as pd 
import numpy as np

from matplotlib import pyplot as plt
from sklearn.preprocessing import StandardScaler

import ipympl 
%matplotlib ipympl


file_path = 'ANN_2022_predictions.csv'

df = pd.read_csv(file_path)

df.set_index('ts', inplace=True)

df.index = pd.to_datetime(df.index)

plt.figure(figsize=(22,10))

plt.plot(df["Real kW"], label="Real kW", color="blue")
plt.plot(df["Predicted kW"], label="Predicted kW", color="red")
plt.grid(True)

# Adding labels and title
plt.xlabel("Index")
plt.ylabel("kW")
plt.title("Real kW vs Predicted kW")

# Adding legend
plt.legend()

# Displaying the plot
plt.show()

```
